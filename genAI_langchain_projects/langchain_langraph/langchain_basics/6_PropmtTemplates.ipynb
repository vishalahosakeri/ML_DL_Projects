# 1. PromptTemplate :Prompt template for a language model
 a. Intialization from PromptTemplate directly <br>
 b. Intialization from from_tempalte

optional_variables doesn't make a value optional at .format() time
 

```python
from langchain_core.prompts import PromptTemplate

# Intialization from PromptTemplate directly
template = PromptTemplate(input_variables=["name","message"],partial_variables={"age":"unknown"},template="Hi, I am {name},age {age}. Giving message : {message}")
message = template.format(name = "Vikram",message = "I am good here." )
# template.
print(message,type(message))
```

.invoke() is the LCEL-standard way to call it (since PromptTemplate is a Runnable), and it behaves a little differently from .format()
.format is not chainable.

```python
message = template.invoke({"name":"Vikram","age":37, "message":"Dont worry about me, I am doing good."}) #lass 'langchain_core.prompt_values.StringPromptValue'
print(message.text,type(message))
```

```python
template = PromptTemplate.from_template(template = "Hi, I am {name},age {age}. Giving message : {message}")
template = template.partial(age=20)
print(template.invoke({"name":"Vishala","message":"I miss you."}))
```

# 2. ChatPromptTemplate :Prompt template for chat models.

```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate,HumanMessagePromptTemplate,AIMessagePromptTemplate

messages = [
    ("system","You are helpful assistant for ai learning.Your name is {name}"),
    ("human","Hello,How are you doing?"),
    ("ai","Doing good, thanks! How can I help you today?"),
    ("human","{userInput}")
]

chatTemplate = ChatPromptTemplate(messages=messages)
tempalteMsg = chatTemplate.invoke({"name":"Vikram","userInput":"What is weight decay?"})
print(tempalteMsg)

test =ChatPromptTemplate.from_template(template="Hi")
print(test.invoke({}))

chatTemplate = ChatPromptTemplate.from_messages(messages=messages)
tempalteMsg = chatTemplate.invoke({"name":"Vikram","userInput":"What is weight decay?"})
print(tempalteMsg)

chatTemplate = ChatPromptTemplate.from_messages([SystemMessagePromptTemplate.from_template("You are helpful assistant for ai learning.Your name is {name}."),
                                                 HumanMessagePromptTemplate.from_template("{userInput}")])
tempalteMsg = chatTemplate.invoke({"name":"Vikram","userInput":"What is weight decay?"})
print(tempalteMsg)
```

# Few shots:
8. FewShotPromptTemplate :Prompt template that contains few shot examples.
9. FewShotChatMessagePromptTemplate :Chat prompt template that supports few-shot examples.

```python
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate

examples = [
    {"input":"bad","output":"good"},
    {"input":"tall","output":"short"},
    {"input":"In","output":"out"}
]

base_tempalte = PromptTemplate(input_variables=["input","output"],template="Input:{input}\nOutput:{output}")
fewShotTemplate = FewShotPromptTemplate(examples = examples, example_prompt = base_tempalte, prefix = "Give the antonym for every input.",
                                        suffix = "input:{adjective}\noutput:",input_variables = ["adjective"])
print(fewShotTemplate.invoke({"adjective":"big"}).to_string())
```

```python
from langchain_core.prompts import ChatPromptTemplate, FewShotChatMessagePromptTemplate

examples = [
    {"input":"bad","output":"good"},
    {"input":"tall","output":"short"},
    {"input":"In","output":"out"}
]

example_prompt = ChatPromptTemplate.from_messages([
    ("human","{input}"),
    ("ai","{output}")
])

fewShot_template = FewShotChatMessagePromptTemplate(examples=examples,example_prompt=example_prompt)

final_prompt = ChatPromptTemplate.from_messages([
    ("system","You are designed for giving opposite words."),
    fewShot_template,
    ("human","{adjective}")
])

print(final_prompt.invoke({"adjective":"big"}).to_messages())
```

# Dynamic example selection

Quick trace of what happens<br>
1. .invoke({"adjective": "small"}) is called.
2. Since example_selector is set (not examples directly), LangChain calls example_selector.select_examples({"adjective": "small"}) — this embeds "small" and does a FAISS similarity search to pull the k=2 closest examples from your 3.
3. Each selected example gets formatted through example_prompt ("input:{input}\nouput:{output}").
4. Everything gets assembled: prefix → selected examples → suffix (with {adjective} filled in).

```python
from langchain_core.example_selectors import SemanticSimilarityExampleSelector
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_classic.vectorstores import FAISS

examples = [
    {"input":"bad","output":"good"},
    {"input":"tall","output":"short"},
    {"input":"In","output":"out"}
]

example_prompt = PromptTemplate(input_variables=["input","output"],template="input:{input}\nouput:{output}")

embeddings = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
example_selector = SemanticSimilarityExampleSelector.from_examples(examples=examples,embeddings=embeddings,vectorstore_cls=FAISS,k=2)
fewShot_template = FewShotPromptTemplate(example_selector=example_selector, example_prompt=example_prompt,prefix="Display opposite words.",suffix="input:{adjective}\noutput:",input_variables=["adjective"] )
message = fewShot_template.invoke({"adjective":"small"})
print(message)
```

# Partial prompt templates — pre-binding some variables <br>
Two use cases: <br>
(a) you know a value early and want to reuse the template, or <br>
(b) the value comes from a function evaluated at call time (e.g. current date) <br>


```python
# a.
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(input_variables=["adjective","content"],template="Tell me a {adjective} about {content}",partial_variables={"adjective":"funny"})
print(prompt.invoke({"content":"AI engineers"}))
```

```python
#b
from datetime import datetime

def get_date():
    return datetime.now().strftime("%y-%m-%d")
prompt = PromptTemplate(input_variables=["thought","todayDate"],template="Today's date is {todayDate},thought of the day:{thought}",partial_variables={"todayDate":get_date})
print(prompt.invoke({"thought":"Be brave"}))
```
