# parallel execution

```python
from typing import TypedDict

class EmpState(TypedDict):
    name : str
    montly_salary : float
    yearly_salary : float
    bonus : float
    yearEndAmmt : float

    
```

```python
from langgraph.graph import StateGraph, START ,END

# define nodes

def calculate_bonus(state : EmpState):
    bonus = state['montly_salary'] * 0.2
    return {
        "bonus" : bonus
    }

def calculate_yearSalary(state : EmpState):
    yearly_salary = state['montly_salary'] * 12
    return {
        "yearly_salary" : yearly_salary
    }

def calculate_yearEndAmt(state : EmpState):
    yearEndAmmt = state['bonus'] + state["yearly_salary"]
    return {
        "yearEndAmmt" : yearEndAmmt
    }

# define graph

graph = StateGraph(state_schema=EmpState)

graph.add_node("calculate_bonus",calculate_bonus)
graph.add_node("calculate_yearSalary",calculate_yearSalary)
graph.add_node("calculate_yearEndAmt",calculate_yearEndAmt)

# define edges
graph.add_edge(START, "calculate_bonus")
graph.add_edge(START, "calculate_yearSalary")
graph.add_edge("calculate_bonus", "calculate_yearEndAmt")
graph.add_edge("calculate_yearSalary", "calculate_yearEndAmt")

workflow = graph.compile()

initial_state = {"name":"vsh","montly_salary":2000}
final_state = workflow.invoke(initial_state)
```
