```python
import torchmetrics
import mlxtend
import torch

from torchmetrics import ConfusionMatrix,Accuracy,Precision,Recall,F1Score
from mlxtend.plotting import plot_confusion_matrix
import numpy as np

y_actual = torch.tensor([1,1,1,1,0])
y_pred =   torch.tensor([1,0,1,0,1])
cnfMatrix = ConfusionMatrix(task='binary',num_classes=2)
cnfMatrix = cnfMatrix(y_pred,y_actual)
print(cnfMatrix)
acc = Accuracy(task="binary",num_classes=2)
print("accuracy:",acc(y_pred,y_actual))
precison = Precision(task="binary",num_classes=2)
print("precision:",precison(y_pred,y_actual))
recall = Recall(task="binary",num_classes=2)
print("recall:",recall(y_pred,y_actual))
plot_confusion_matrix(cnfMatrix.numpy())



y_actual = torch.tensor([1,2,3,1,0,3])
y_pred =   torch.tensor([1,2,1,1,0,2])
cnfMatrix = ConfusionMatrix(task='multiclass',num_classes=4)
cnfMatrix = cnfMatrix(y_pred,y_actual)
print(cnfMatrix)
acc = Accuracy(task="multiclass",num_classes=4)
print("accuracy:",acc(y_pred,y_actual))
precison = Precision(task="multiclass",num_classes=4)
print("precision:",precison(y_pred,y_actual))
recall = Recall(task="multiclass",num_classes=4)
print("recall:",recall(y_pred,y_actual))
plot_confusion_matrix(cnfMatrix.numpy())

f1 = F1Score(task="multiclass",num_classes=4)
print("f1-score:",f1(y_pred,y_actual))

```
