# get the data to practice BoW

```python
import download_file
from download_file import downloadFile
import pandas as pd

url = "https://raw.githubusercontent.com/sumankumarsubudhi/Spam-Classifier/master/SMSSpamCollection"
output_file = "SMSSpamCollection.txt"

downloadFile(url,output_file)

messages = pd.read_csv("SMSSpamCollection.txt", sep ='\t',names=["label","message"])

print(messages)
```

```python
# Process the text for getting BoW
import nltk
from nltk import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
import sklearn
from sklearn.feature_extraction.text import CountVectorizer
import re

# nltk.download('wordnet')
# nltk.download('stopwords')

stpWords = stopwords.words('english')
leammatizer = WordNetLemmatizer()

corpus = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [leammatizer.lemmatize(word) for word in review if word not in set(stpWords)]
    review = " ".join(review)
    corpus.append(review)
    
print(corpus)

```

# Tradeoff

Using larger n-grams:<br>

captures more context<br>
but increases vocabulary size<br>
increases sparsity

unigrams → small vocab<br>
bigrams → larger vocab<br>
trigrams → very large vocab

```python
#get BoW

cntVector = CountVectorizer(max_features=100,ngram_range=(2,2))
X = cntVector.fit_transform(corpus).toarray()
print(X.shape)
print(X)
```

# TF-IDF 

TF = nuber of repetition of word / number of words in sentence
IDF = log(numer of sentences/ nubmer of sentences containing that word)

final vector of sentences = TF * IDF

Advantages -> intuitive, fixed size input, word importance 
disadvnt -> oov, sparsity 

```python
import sklearn
from sklearn.feature_extraction.text import TfidfVectorizer

tf = TfidfVectorizer(max_features=10,ngram_range=(1,1))
X = tf.fit_transform(corpus).toarray()

print(X)
print(X.shape)
```
