1. RecursiveCharTextSplitter -priority order: ["\n\n", "\n", " ", ""]. It tries to split on paragraphs first; if a resulting chunk is still too big, it recurses into that chunk and tries the next separator (sentences → words → characters)
2. CharTextSplitter  - Splits on a single separator (default "\n\n")
3. HTMLHeaderTextSplitter - header , split text from file

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("/Users/sanvijanvi/Documents/SJ_HomeWork/speech.txt")
text = loader.load()
```

```python
from langchain_text_splitters import CharacterTextSplitter

splitter = CharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
chunks = splitter.split_text(text[0].page_content)
print(type(chunks),len(chunks))
print(type(chunks[0]), chunks[0])
print(chunks[1])
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(file_path="/Users/sanvijanvi/Documents/SJ_HomeWork/attention.pdf")
pdfDocs = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

seperator = ["\n\n","\n"," ",""]
splitter = RecursiveCharacterTextSplitter( chunk_size = 1000, chunk_overlap = 200, separators=seperator)
pdfChunks = splitter.split_documents(pdfDocs)
print(type(pdfChunks),type(pdfChunks[0]))
print(pdfChunks[0].page_content)
print("*******************************")
print(pdfChunks[1].page_content)
```

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

header = [
    ("h1","header1"),
    ("h2","header2"),
    ("h3","header3")
]

splitter = HTMLHeaderTextSplitter(headers_to_split_on=header)
htmlChunks = splitter.split_text_from_file("/Users/sanvijanvi/Documents/SJ_HomeWork/example.html")
print(len(htmlChunks),type(htmlChunks),type(htmlChunks[0]))
# print(htmlChunks[0].page_content)

for chunk in htmlChunks:
    print(chunk.page_content)
```
