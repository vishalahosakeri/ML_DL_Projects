```python
sentence = "Cat say meow. Dog say bark."

import nltk
from nltk.tokenize import word_tokenize, sent_tokenize, wordpunct_tokenize

sent = sent_tokenize(sentence)
wordList = []

for s in sent:
    word = wordpunct_tokenize(s)
    cleanWords = [w for w in word if w.isalnum()]
    wordList.append(cleanWords)

wordList
```

```python
from gensim.models import Word2Vec
import numpy as np

model = Word2Vec(wordList,vector_size=5,min_count=1,window=5)

cat_vector = model.wv["Cat"]
v_val = np.array([model.wv["meow"],model.wv["bark"],model.wv["Dog"]])

scores = model.wv.cosine_similarities(cat_vector,v_val)
print(scores)

model.wv.get_index("Dog")
model.wv.get_vector("Cat")
```
