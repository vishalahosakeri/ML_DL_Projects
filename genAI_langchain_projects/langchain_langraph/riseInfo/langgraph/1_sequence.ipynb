# sequence flow 

Temperature conversion and label

```python
from typing import TypedDict,Literal

class TempConversionState(TypedDict):
    temp_cent : float
    temp_farn : float
    label : Literal["very hot","hot","warm","cold"]
```

```python
from langgraph.graph import StateGraph, START,END

# define the nodes
def tempConv(state : TempConversionState):
    print("Node1 : tempConv ")
    tempCent = (state["temp_farn"] - 32) * 5/9
    return {
        "temp_cent" : tempCent
    }

def tempLabel(state : TempConversionState):
    print("Node2 : tempLabel ")
    tempCent = state["temp_cent"]
    if tempCent > 30 :
        label = "very hot"
    elif tempCent < 30 and tempCent >25:
        label = "hot"
    elif tempCent <= 25 and tempCent >15:
        label = "warm"
    else :
        label = "cold"
    return {
        "label" : label
    }

# define graph
graph = StateGraph(state_schema=TempConversionState)

# add nodes
graph.add_node("tempConv",tempConv)
graph.add_node("tempLabel",tempLabel)

# add edges
graph.add_edge(START, "tempConv")
graph.add_edge("tempConv","tempLabel")
graph.add_edge("tempLabel",END)

workflow = graph.compile()

intial_state = {"temp_farn":10}
final_state = workflow.invoke(input = intial_state)
print(final_state)
```
