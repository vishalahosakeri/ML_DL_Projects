# Get dataset

```python
import torch
import torchvision
from torchvision import datasets
import matplotlib.pyplot as plt

train_data = datasets.FashionMNIST(root="data",train=True,download=True,transform=torchvision.transforms.ToTensor())
test_data = datasets.FashionMNIST(root="data",train=False,download=True,transform=torchvision.transforms.ToTensor())

len(train_data), len(test_data)
```

```python
image, label = train_data[0]
plt.imshow(image.squeeze(),cmap="gray")
plt.title(label)
```

```python
print("class_to_idx:",train_data.class_to_idx)
print(train_data.classes)
print(train_data.targets)

classes = train_data.classes
labels = train_data.train_labels
print(labels)
print(classes[6])

img, label = train_data[1]
plt.imshow(img.squeeze(),cmap="gray")
plt.title(classes[label])
```

```python
#print random 16 images
torch.manual_seed(42)
rows, cols = 4, 4

fig = plt.figure(figsize=(9,9))
for i in range (1,rows*cols+1):
    # print(i)
    rnd_indx = torch.randint(0,len(train_data),size=[1]).item()
    # print(i)
    img, label = train_data[rnd_indx]
    fig.add_subplot(rows,cols,i)
    plt.imshow(img.squeeze(),cmap="gray")
    plt.title(classes[label])
    plt.axis(False)

```
