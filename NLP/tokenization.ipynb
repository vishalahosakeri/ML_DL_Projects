# STEMMING

Porter is the original English-only stemmer. 
Snowball is the improved, multilingual upgrade — use Snowball unless you have a specific reason for Porter.

SnowballStemmer("spanish").stem("corriendo")  # → corr

```python
from nltk.stem import SnowballStemmer, RegexpStemmer, PorterStemmer

words = ["eat","eating","eaten","eats","stopping","careful","History","Bully","Silly","Congratulations"]
stemmerList = []

portStemmer = PorterStemmer()
stemmerList.append(portStemmer)
sbStemmer = SnowballStemmer(language="english")
stemmerList.append(sbStemmer)
regStemmer = RegexpStemmer('ing$|s$|e$|able$',min=4)
stemmerList.append(regStemmer)

for stemmer in stemmerList:
  print(f"Printing for deffierent stemmer - {stemmer.__class__}")
  for i in range(len(words)):
    print(f"{words[i]} --> {stemmer.stem(words[i])}")
```

# Lemmitization

Lemmatization only reduces a word to its base dictionary form


# Core Difference

1. Stemming — crudely chops off word endings using rules. Result may not be a real word.
2. Lemmatization — intelligently reduces to the actual dictionary base form. Always a real word.

# When to Use Which
Use Stemming when:<br>

1. Speed matters more than accuracy
2. Large scale search engines / information retrieval
3. You don't need real words in output

Use Lemmatization when:<br>

1. Accuracy matters
2. Sentiment analysis
3. Chatbots / text understanding
4. Output needs to be human readable<br>


One Line Summary <br>

Stemming is a fast but dumb chopper. Lemmatization is a slow but smart dictionary lookup.

```python
from nltk.stem import WordNetLemmatizer
words = ["eat","eating","eaten","eats","stopping","careful","History","Bully","Silly","Congratulations","better","worst"]

nltk.download("wordnet")
lemitizer = WordNetLemmatizer()
listPos = ["n","v","a","r"]

for pos in listPos:
 print(f"***** {pos} ****")
 for word in words:
  print(f"{word} --> {lemitizer.lemmatize(word,pos=pos)}")
```

# POS tags

Part-of-Speech (POS) tagging in Natural Language Processing (NLP) is the process of automatically assigning grammatical categories (like nouns, verbs, or adjectives) to each word in a sentence. <br>

POS tagging acts as a foundational building block for more complex NLP applications:<br>
1. Named Entity Recognition (NER): Identifies proper nouns to extract names, locations, and organizations.
2. Machine Translation: Ensures translated words adhere to the correct grammatical structures of the target language.
3. Text-to-Speech (TTS): Helps decide the correct pronunciation of words (e.g., knowing “object” is a verb versus a noun).

# tree bank => Need full understanding

The Treebank corpus in NLTK is a syntactically annotated collection of text that provides a full syntactic parse (parse tree) for each sentence. It is primarily used for training and evaluating part-of-speech taggers and parsers
<br>
NLTK includes samples of major treebanks, as full versions often require paid licenses: <br>
1. Penn Treebank (treebank): A 10% sample of the full Penn Treebank, containing approximately 3,914 sentences.
2. Sinica Treebank (sinica_treebank): A sample of 10,000 parsed Chinese sentences from the Academia Sinica Balanced 
3. Corpus.PropBank (propbank): An augmentation of the Penn Treebank that includes predicate-argument structure annotations for verbs.

```python
import nltk
from nltk.corpus import treebank

nltk.download("treebank")
treebank.words()

# Access the first parsed sentence
sentence = treebank.parsed_sents()[0]
print(sentence)

# Access tagged words from a specific file
tags = treebank.tagged_words('wsj_0001.mrg')
print(tags[:5])
```

# Named entity recognition

NER identifies and classifies named entities in text into categories such as:<br>

Person → "Sachin Tendulkar"<br>
Organization → "Google", "OpenAI"<br>
Location → "Bangalore", "India"<br>
Date/Time → "June 2026"<br>
Money → "$1000"<br>
Percentage → "15%"

```python
from nltk import word_tokenize, sent_tokenize, ne_chunk, ne_chunk_sents, pos_tag

paragraph = "I have three visions for India. In 3000 years of our history, people from all over the world have come and invaded us, captured our lands, conquered our minds.Apple was founded by Steve Jobs in Cupertino in 1976."

nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
sentences = sent_tokenize(paragraph)

for sent in sentences:
    wordsToken = word_tokenize(sent)
    word_tags = pos_tag(wordsToken)
    entities = ne_chunk(word_tags)
    entities.draw()
    print(entities)
```
