```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("/Users/sanvijanvi/Documents/SJ_HomeWork/speech.txt")
text = loader.load()
print(type(text), len(text))
print(type(text[0]))
print("Page content:",text[0].page_content)
print("Meta data:",text[0].metadata)
```

# pdf loader

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(file_path="/Users/sanvijanvi/Documents/SJ_HomeWork/attention.pdf")
pdfDocs = loader.load()
print(len(pdfDocs))
print(pdfDocs[2].page_content)
```

DirectoryLoader

```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader


directoryLoder = DirectoryLoader(path = "//Users/sanvijanvi/Documents/SJ_HomeWork",glob= "*.pdf", loader_cls=PyPDFLoader, silent_errors=True)
pdfDocs = directoryLoder.load()

print(len(pdfDocs))
```

# PyDirectory laoder

```python
from langchain_community.document_loaders import PyPDFDirectoryLoader

pyDirectoryLoader = PyPDFDirectoryLoader(path = "//Users/sanvijanvi/Documents/SJ_HomeWork",mode='single')
pdfDocs = pyDirectoryLoader.load()
print(len(pdfDocs),type(pdfDocs))

for doc in pdfDocs:
    print("Metadata :",doc.metadata["creator"])
```
