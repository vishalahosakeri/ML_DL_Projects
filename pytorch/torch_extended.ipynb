# reshape vs view

Feature             view()                                             reshape()
Data Sharing   Always returns a view (shares the same data).     Returns a view when possible; otherwise returns a copy.Contiguity    Requires the tensor to be contiguous in memory.      Can handle both contiguous and non-contiguous tensors.
Error Handling  Throws a RuntimeError if the tensor is non-contiguous.   Automatically copies data if needed to fulfill the new shape.

Note:
The outputs look identical — the real difference is about memory, not the result.The outputs are identical — the difference is invisible in print but very real under the hood.

Practical rule for nn: just use reshape everywhere. Use view only when you intentionally want to share memory and catch non-contiguous bugs early. In 99% of model code you'll see reshape.

```python
mat1 = torch.arange(1,10).reshape(3,3)
print(mat1)

mat1 = mat1 + 3
print(mat1)
```

```python
mat1 = torch.arange(1,10).view((3,3))
print(mat1)

mat1 = mat1 + 3
print(mat1)
```

```python
# proof of non - contiguous
t = torch.arange(1,5).reshape(2,2)
v = t.view((4))

t[0,0]= 9   # upating t is updating v
print(t)
print(v)

#transpose creates non-contighuous
t1 = t.T
print(t1)
print(t1.reshape(4))
print(t1)
print(t1.contiguous().view(4)) # if we dont use contiguous, throws error - view size is not compatible with input tensor's size and stride (at least one dimension spans across two contiguous subspaces). Use .reshape(...) instead.
```

# permute, transpose

Difference between T and transpose:
shape [2, 3, 4] => t3.T     =>  reverses ALL dims → shape [4, 3, 2]
torch.transpose(t3, 0, 1)     # swaps ONLY dim0 and dim1 → shape [3, 2, 4]
torch.transpose(t3, 1, 2)     # swaps ONLY dim1 and dim2 → shape [2, 4, 3]

.T / .transpose()
    → no copy
    → same memory
    → strides flipped
    → is_contiguous() = False
    → view() will ERROR
    → reshape() works (makes a copy internally when needed)

.contiguous()
    → forces a real copy
    → reorders memory to match strides
    → is_contiguous() = True
    → view() works again

t.T.is_contiguous()              # False  — no copy yet
t.T.contiguous().is_contiguous() # True   — NOW a copy was made
t.T.reshape(6)                   # copy made silently inside reshape
t.T.view(6)                      # ERROR  — refuses to copy

Note : Only .clone(), .contiguous(), or reshape (when forced) actually allocate new memory.


```python
mat1 = torch.arange(1,25).reshape(2,3,4)
print(mat1, mat1.shape)

# T
# print(mat1.T)

#transpose
print(torch.transpose(mat1,0,1))#shape (3,2,4)
```

# permute

# Broadcasting rules -> Important

Rule 1 — align shapes from the RIGHT, pad 1s on the left
Rule 2 — a size-1 dim expands to match the other tensor's size
Rule 3 — if neither dim is 1 and they differ → ERROR

```python
#RULE 1  align shapes from the RIGHT, pad 1s on the left

t1 = torch.ones(3,1) # shape (3,1)
t2 = torch.ones(4)   # shape(1,4) -> rule applied here

print(f"t1 shape:{t1.shape},t2 shape : {t2.shape}")
t3 = t1 + t2
print(t1,t2)
print(t3,t3.shape) # shape of t3 = (3,4)

#Rule 2 — a size-1 dim expands to match the other tensor's size
t1 = torch.ones(3,1)
t2 = torch.ones(1,4)
t3 = t1+t2
print(t3,t3.shape) # shape of t3 = [3, 4]  ← 3×1 expands to 3×4

#Rule 3 — if neither dim is 1 and they differ → ERROR
t1 = torch.ones(3,2)
t2 = torch.ones(1,4)
t3 = t1+t2 # RuntimeError: The size of tensor a (2) must match the size of tensor b (4) at non-singleton dimension 1
print(t3,t3.shape) # shape of t3 = [3, 4]  ← 3×1 expands to 3×4
```

Expaliantion:
Step 1 — Rule 1 kicks in first (pad 1s on the left):
[32, 10]
[    10]  ← shorter, so pad left with 1 → [1, 10]
Step 2 — Rule 2 kicks in (size-1 expands):
[32, 10]
[ 1, 10]  ← dim0 is 1, expands to 32
─────────
[32, 10]  ← result
Now your confusion — why does [3,2] + [1,4] fail but [32,10] + [10] work?
[3, 2]        [32, 10]
[1, 4]        [ 1, 10]  ← after padding

dim0: 3 vs 1  ✅ (one is 1, expands)    dim0: 32 vs 1  ✅ (one is 1, expands)
dim1: 2 vs 4  ❌ (neither is 1, ERROR)  dim1: 10 vs 10 ✅ (equal, no expand needed)
The key difference is dim1:

2 vs 4 → neither is 1, not equal → ERROR
10 vs 10 → equal → fine

So the rule to check for any dim pair is simply:
dim_a == dim_b        ✅ equal — no expand needed
dim_a == 1            ✅ a expands to match b
dim_b == 1            ✅ b expands to match a
none of the above     ❌ ERROR
