1. BS4 - beautifulSoup

```python
from bs4 import BeautifulSoup
import requests

def scrape_url(url):
    response = requests.get(url=url,headers={"User-Agent": "Mozilla/5.0"},timeout=10)
    response.raise_for_status()

    soup = BeautifulSoup(markup=response.text, features="html.parser")

    for tag in soup(["style","nav","script","header","footer"]):
        tag.decompose()

    text = soup.get_text(separator="\n",strip=True)
    return text

print(scrape_url("https://timesofindia.indiatimes.com/entertainment/hindi/movie-reviews/dhurandhar/movie-review/125786542.cms"))
```

```python
from ddgs import DDGS

def web_search(query:str)->str:
    '''Search the web with given query for up to date information.Return titles ,url ansd snippets. '''
    try :
        results = DDGS().text(query,max_results = 5)
    except Exception as e:
        return f"Search failed:{e}"
    print(results)
    return " ".join([f"Title:{r['title']}\nUrl:{r['href']}\nSnippet:{r['body']}" for r in results])

print(web_search(query="movie review about Dhurandar"))
```
