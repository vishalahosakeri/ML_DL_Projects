# Difference between mmul and mm
Feature                 torch.mm                                torch.matmul
Broadcasting             No                                     Yes
Supported Dimensions     Strictly 2D                            1D, 2D, and higher
Batch Multiplication     No                                     Yes (via broadcasting)
Operator Alias           None                                   The @ operator in Python 3

mm when you want to ensure your inputs are strictly matrices and avoid accidental broadcasting errors. Use matmul for more flexible operations, especially when working with batches or vectors

```python
mat1.mul(mat2)  #elementwise multiplication
mat1.mm(mat2) # dot product
mat1.matmul(mat2) # dot product

vec1 = torch.tensor([1, 2, 3])
vec2 = torch.tensor([4, 5, 6])

# This works and returns 32
result = torch.matmul(vec1, vec2) 
print(f"matmul: {result}")

print(mat1.mm(mat2) ,mat1.matmul(mat2))

# This fails and raises a RuntimeError
result = torch.mm(vec1, vec2) 
print(f"mm: {result}")
```

# Unique - print unique values preserving original order



```python

t4 = torch.tensor([1,8,3,1,2])
uniqVal,inverse,cnt = torch.unique(t4,sorted=True,return_inverse=True,return_counts=True)
print(uniqVal,inverse,cnt)
print("original:",uniqVal[inverse])
uniqeElemCnt = inverse.max().item()+1
print(uniqeElemCnt)
first_occurance = torch.zeros(uniqeElemCnt,dtype=torch.long)

for i in range(len(t4)-1,-1,-1):
    first_occurance[inverse[i]]=i
    print(inverse[i],first_occurance[inverse[i]],i)

insertion_order = first_occurance.sort().values

print(first_occurance)
print(insertion_order)
print(t4[insertion_order])
```
