# Single agent system 

-> get movie review from web search

1. llm
2. tools

```python
from langchain_groq.chat_models import ChatGroq
import os 
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(model = "openai/gpt-oss-20b", max_tokens=1024)
```

```python
from langchain_core.tools import tool
from ddgs import DDGS

@tool
def webSearch_movie(movieName : str):
    ''' This tool searches the web for given query which is movie name and send the results in the format of title, url and review'''
    try :
        result = DDGS().text(movieName, max_results=3)
    except Exception as e :
        print(f"Web search has issue {e}")
    return result

# answer = webSearch_movie.invoke("Please give the review about movie Dhurandar")

# for r in answer:
#     print(f"title : {r['title']}, url : {r['href']},review :{r['body']}")
```

```python
from langchain.agents import create_agent

tools = [webSearch_movie]
agent = create_agent(model=llm, tools=tools)

result = agent.invoke(input={"messages":[{"role":"user","content":"Please give the review about movie Dhurandar"}]})
```
